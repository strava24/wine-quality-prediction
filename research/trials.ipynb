## Explanation on what is ConfigBox

 ConfigBox comes from the python-box library. It's a dict subclass that lets you read keys as attributes. It also converts nested dicts into the same type automatically.

```python
d1 = {"key1": "value1", "key2": "value2"} # define a dict
```

```python
# in order to access the values the way is
d1["key1"]
```

```python
# You cannot simply do something like
d1.key1
```

```python
from box import ConfigBox
d2 = ConfigBox(d1)

d2.key1 # this is acceptable in checkbox
```

## Explanation on what is @ensure_annotations decorator
@ensure_annotations is a decorator from the ensure library. It checks the type hints on a function at runtime and raises an error if they're violated.

 Python's type hints are normally not enforced. This works without any complaints. With the decorator, the call is checked before the function body runs

```python
def calculate_product(x: int, y: int) -> int:
    return x * y

calculate_product(2, 6)  # this works without any issues
```

```python
# look into this
calculate_product(2, "6")
# even though the parameter type is defined in the method declaration
# this still works without any issues
```

```python
# now lets try that with @ensure_annotations

from ensure import ensure_annotations

@ensure_annotations
def calculate_ensure_product(x: int, y: int) -> int:
    return x * y

calculate_ensure_product(2, 6) # right scenario
```

```python
calculate_ensure_product(2, "6") # wrong scenario
```
